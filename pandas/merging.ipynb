```python
import pandas as pd
```

```python
df1 = pd.read_csv(r"LOTR.csv")
df1
```

```python
df2 = pd.read_csv(r"LOTR 2.csv")
df2
```

```python
#left     right
# df1.merge(df2) # inner is by default
df1.merge(df2 , how = "inner" , on= ["FellowshipID","FirstName"])
```

```python
df1.merge(df2 , how = "outer" , on= ["FellowshipID","FirstName"])
```

```python
df1.merge(df2 , how = "left" , on= ["FellowshipID","FirstName"])
```

```python
df1.merge(df2 , how = "right" , on= ["FellowshipID","FirstName"])
```

```python
df1.join(df2,on ="FellowshipID",how = "outer",lsuffix = "_left",rsuffix = "_right")
```

```python
df4 = df1.set_index("FellowshipID").join(df2.set_index("FellowshipID"),on ="FellowshipID",how = "left",lsuffix = "_left",rsuffix = "_right")
# df1.set_index("FellowshipID")
```
