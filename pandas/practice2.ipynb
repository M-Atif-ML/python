```python
new_data= {"Column_Name":[],"Number_of_missing_values":[],"Percentage":[]}
df2 = pd.DataFrame(new_data)
```

```python
df2["Column_Name"]=list(df.columns)
```

```python
df2=df2.set_index("Column_Name")
```

```python

for x in list(df.columns):
    df2.loc[x,"Number_of_missing_values"]=106077-df[x].count()
```

```python
# df2=df2.reset_index()
for x in list(df2.index):
    df2.loc[x,"Percentage"]=df2.loc[x,"Number_of_missing_values"]/106077*100
```

```python
df2
```

```python
df2 = df2.sort_values("Percentage",ascending=False)
```

```python
for x in df2.index:
    if df2.loc[x,"Percentage"] > 60:
        df.drop(columns = x,inplace =True)
```
