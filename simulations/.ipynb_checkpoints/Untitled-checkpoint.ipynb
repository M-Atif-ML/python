```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python

df = pd.DataFrame({
    "Diameter": [8, 10, 12],
    "Price": [10, 13, 16]
})
```

```python
mean_x,mean_y= df.mean(axis= 0 )["Diameter"],df.mean(axis= 0 )["Price"]
```

```python
mean_x,mean_y
```

```python
df["Deviations(x)"]=df["Diameter"]-mean_x
df["Deviations(y)"]=df["Price"]-mean_y
```

```python
df["Product_of_dev"] = df["Deviations(x)"]*df["Deviations(y)"]
```

```python
df
```

```python
sum_product_dev = df.sum(numeric_only=True)["Product_of_dev"]
sum_product_dev
```

```python
df["Square_of_dev_x"]=df["Deviations(x)"]**2
```

```python
df
```

```python
sum_sq_of_dev_x= df.sum(numeric_only=True)["Square_of_dev_x"]
```

```python
slope = sum_product_dev/sum_sq_of_dev_x
slope
```

```python
x = int(input("Enter Diameter: "))
b = mean_y - slope * mean_x
output = slope*x+b
output
```
